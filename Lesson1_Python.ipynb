### Vezérési szerkezetek, fori, foreach, if-elif-else, while (break, continue)

```python
for i in range(5):  # 0.., 5 exclusive # for (int i = 0; i < 5; ...
    print(i, end = '') # 01234
print() # newline

for i in range(1, 5, 2):  # for (int i = 1; i < 5, i += 2) ...
    print(i) # 1, 3

for animal in ["dog", "cat", "mouse"]: # foreach
    print (animal)

animals = ["dog", "cat", "mouse"]
for i, animal in enumerate(animals): # fori + foreach
    print (f"{i}. {animal}")


numbers = [1, 2, 3]
for x, a in zip(numbers, animals): # két, azonos hosszú lista
    print(f'{x} {a}')

for i in range(4, 8): 
    if i == 3: # if-elif-else
        print(f'broke at 3')
        break
    elif i == 2:
        print(f'broke at 2')
        break
    else: 
        print("else")
        continue

count = 0
while count < 5: # while 
    print(count)
    count += 1 # x = x + 1
```

### Listák

```python
# listák (== tömb)
x = [1,2,3]

x.append(4) # végéhez fűz
print('pop:')
print(x.pop()) # végéről eltávolít (és ki is írja)

print(x[2]) # 0-val kezdődik az indexelés
print(x[-1]) # -1 utolsó elem ([-2] utolsó előtti)
print(x[1:3]) # első index inkluzív, a második exkluzív, 

print()

x[1:] # slicing, sublist
print(x) # nem modosit
z = x[:2]
print(f'z: {z}')
print(x[::2]) # x[start:end:step] , itt semmit sem irtunk a start, end-re, 2-őt lép és így megall a 3ason (utso elem)

del x[2] # 2-es indexű elem eltávolítása
x.remove(2) # az (első) 2 értékű elem eltávolítása
x.insert(1, 3) # 1-es indexű pozícióra beszúrás

# rendezés
y = sorted(x)
y.reverse() # in-place
print(y)
z = list(reversed(x)) # copy

# min, max, length
print(max(z))
print(min(z))
print(len(z))
```

### List comprehension: [expression `for` item `in` iterable `if` condition]

```python
# list comprehension: [expression for item in iterable if condition]

k = []
for i in [1, 2, 3]:
    u = 10 + i # nem kell kivul inicializalni
    k.append(u)
print(k)
#ua.:
x = [10 + i for i in [1, 2, 3]] # => [11, 12, 13]
print(x)

p = [x for x in [3, 4, 5, 6, 7] if x > 5] # => [6, 7] (kell az elejere is x!!)
print(p)

squares = [x**2 for x in range(5) if x % 2 == 0]  # => [0, 4, 16]
print(squares)
# range 0-tól, 5 exclusive, csak párosok, négyzeteik
```

### Lambda függvények: `lambda` paraméter(ek)`:` kifejezés

```python
x = (lambda x: x > 2)(3) # True, fv. deklaráció helyben, és hívás egyből (3) paraméterrel
print(x)

x = (lambda x, y: x ** 2 + y ** 2)(2, 1) # 5
print(x)

# filter(function, iterables) -> iterator, a list() iterál végig
list(filter(lambda x: x > 5, [3, 4, 5, 6, 7])) # [6, 7] filter egy iterátort ad vissza, a list fog ténylegesen végigiterálni

# ua.:
z = [i for i in [3, 4, 5, 6, 7] if i > 5]
print(z)

# map(function, iterables) -> iterator, a list() iterál végig
list(map(max, [1, 2, 3], [4, 2, 9])) # [4, 2, 3]
# a két lista elemeit párhuzamosan hasonlítja össze, és minden pozíción a nagyobbat választja ki
```

### OOP

```python
class Person: # class
    def __init__(self, name, age): # konstruktor
        self.name = name
        self.age = 5
    def greet(self): # első paraméter minden metódusban: self (this)
        return f"Hi, I'm {self.name}!"

p = Person("Alice", 30) # nincs new
print(p.greet()) # Hi, I'm Alice!
```

    $x_{norm} = \frac{x - min}{max-min}$

3. Bináris klasszifikálás:
    $\hat{y}: x_{norm} >= 0.5$

4. Kiíratás

```python
numbers_input = input("Enter integer numbers (comma-separated): ")
nums = numbers_input.split(',')
nums = [int(i) for i in nums]
a = [(i-min(nums))/(max(nums) - min(nums)) for i in nums]
print([f'{i :.2f}' for i in a])

klass = [x >= 0.5 for x in a]
print(klass)
```
