## Mátrixok közötti műveletek

## Normalizálás, MAE, MSE

```python
data = np.array( [7, 3, 1, 2, 9] )
print(data-data.min())
norm = (data - data.min()) / (data.max() - data.min())
print(norm) # 0 és 1 közé skálázza az adatokat

y = np.array( [1, 2, 3] ) # elvárt kimeneti értékek
y_pred  = np.array( [2, 2, 1] ) # prediktált kimeneti értékek

mae = np.mean(np.abs(y - y_pred)) # Mean Absolute Error (MAE) (átlagos abszolut hiba) - a mean atlagolja a matrix elemit
mse = np.mean((y - y_pred)**2) # Mean Squared Error (MSE) - jobban bunteti a nagyobb hibakat
print(mae)
print(mse)
```

## Feladat

Adott $x$ vektor (input-ok) és $y$ vektor (az x-ekhez tartozó, mért kimeneti értékek), mi az optimális (legkisebb négyzetes hiba) $\hat m$ és $\hat b$ lineáris regresszió esetén?

\begin{equation}
\hat m = \frac{\sum x_i y_i  – \bar y \sum x_i} {\sum x_i^2 – \bar x \sum x_i}
\end{equation}

\begin{equation}
\hat b = \frac{\bar y \sum x_i^2  – \bar x \sum x_i y_i} {\sum x_i^2 – \bar x \sum x_i}
\end{equation}

vagy...

\begin{equation}
\hat b = \bar y - \hat m \bar x  
\end{equation}

(Egy kis segítség: $ \bar x = x.mean() $, $ \sum x_i y_i = x @ y $, $ \sum x_i^2 = x @ x $ )
