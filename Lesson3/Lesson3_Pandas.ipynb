## Függő (target) változó, független változók szétválasztása, feldolgozása

```python
df = pd.read_csv("Car details v3.csv") # CSV-ből
y = df['selling_price'] # target
X = df.drop(columns='selling_price') # visszatérési érték a selling_price nélküli dataframe! (inplace=True) - az eredeti dataframet nem bantja
# X sorai az input feature-ök, többdimenziós az input most; y egydimenziós
```

## Érdektelen oszlopok, sorok eldobása

```python
X = X.drop(columns=['name', 'torque']) 
# ha a name csak a gyártó lenne (BMW, Suzuki, ...), nem biztos, hogy el kéne dobni
# torque-ot (nyomaték) azért dobjuk el, mert elég a max_power (lóerő); redundáns?

X = X.drop_duplicates() # nem hasznosak számunkra
X = X.dropna(how='all') # csupa null sorok eldobása
```

## Üres cellák, null (None, NaN) értékek kezelése 

```python
X.isna()[119:121] # True/False mátrix, maszk, True, ha null

X.isna().sum() # hány null van az egyes oszlopokban (True=1-ek összege, False=0)
# year            0
# ...
# mileage         191
# engine          191
# max_power       188
# seats           191


# df = pd.read_csv("processed.cleveland.data.csv", na_values="?") # ?-es cellák null-ként töltődnek fel már beolvasáskor

# X[X['engine'].isna()] # sorok kiiratása, amelyek null-ok adott oszlopban
```

## Opciók null adatok kezelésére

```python
# (0)
# X = X.dropna() # sorok eldobása, ha bármely mező null (None/NaN) => esetleg túl sok adatot eldobhatunk!

# (1) törlés, ha egy adott cella null
X = X.dropna(subset='engine') # sorok eldobása, ha az adott cella null az adott oszolpban

# (2) konstans (0, 1, -1, egyéb?)
X['seats'] = X['seats'].fillna(5) # outlier-eket vezethetünk be, "elhúzhatja" az adatokat - minden null helyre 5-öt ír

# (3) valamilyen számított érték
df['km_driven'] = df['km_driven'].fillna(df['km_driven'].mean()) # null érték helyettesítése átlaggal (ha numerikus)

# X.info()
 # 8   mileage        7907 non-null   object 
 # 9   engine         7907 non-null   object 
 # 10  max_power      7913 non-null   object 
 # 12  seats          7907 non-null   float64

X['seats'] = X['seats'].fillna(X['seats'].mean())
X = X.dropna(subset=['mileage', 'engine', 'max_power']) # eldobjuk a sorokat, ahol ezek null-ok

X.isna().sum() # most már nincs üres (null) adat egyik oszolopban sem
```

## Nem numerikus adatok átalakítása numerikussá

### Mértékegységek, szövegként tartalmazott számértékek

```python
X.info() # object-ek nem jók
 # 2   fuel          6232 non-null   object 
 # 3   seller_type   6232 non-null   object 
 # 4   transmission  6232 non-null   object 
 # 5   owner         6232 non-null   object 
 # 6   mileage       6232 non-null   object 
 # 7   engine        6232 non-null   object 
 # 8   max_power     6232 non-null   object 

# 7 engine, "1248 CC"
X['engine'] = X['engine'].str.replace("CC", "") # str. !
X['engine'] = X['engine'].astype(int)

# 8 max_power, "74 bhp", "103.52 bhp", " bhp"!
X['max_power'] = X['max_power'].str.replace(" bhp", "").str.strip() # .str.strip()
# :((( " bhp" -> "", ezt nem lehet float-ba parse-olni, ez gyakorlatilag null (de nem), törölni kéne
X['max_power'] = X['max_power'].apply(lambda x: x if x != '' else None) # kinullozzuk, aztán töröljük a sort
X = X.dropna(subset='max_power')
X['max_power'] = X['max_power'].astype(float)

# 6 mileage, "23.4 kmpl", "17.3 km/kg"
X['mileage'] = X['mileage'].str.replace("kmpl", "") # str.
X['mileage'] = X['mileage'].str.replace("km/kg", "") # str.
X['mileage'] = X['mileage'].astype(float)

# TODO házi feladat: US magasság->cm? pl. 5’9" = 5* 12*2.54 (cm) + 9* 2.54 (cm) = 175
```


### Enumerációk

```python
print(X['fuel'].unique())

# 4 transmission, "Manual", "Automatic" 
# enum, kétértékű => lehet 0, 1
X['transmission'].unique() # ['Manual', 'Automatic']
X['transmission'] = X['transmission'].apply(lambda x: 1 if x == 'Automatic' else 0) # ternary operátor (? :)

# 3 seller_type 
# enum, de a harmadik érték nem érdekes
df['seller_type'].unique() # ['Individual', 'Dealer', 'Trustmark Dealer']
mapping = {
    'Individual': 0,
    'Dealer': 1,
    'Trustmark Dealer': 1
}
X['seller_type'] = X['seller_type'].map(mapping)

# 5 owner
# enum, de számokkal jól reprezentálható
X['owner'].unique() # ['First Owner', 'Second Owner', 'Third Owner', 'Fourth & Above Owner', 'Test Drive Car']
mapping = {
    'First Owner': 1,
    'Second Owner': 2,
    'Third Owner': 3,
    'Fourth & Above Owner': 4,
    'Test Drive Car': 0
}
X['owner'] = X['owner'].map(mapping)
```

## ...egyéb: join, stack, ...

```python
df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [30, 40]})
df2 = pd.DataFrame({'Name': ['Cecil'],        'Age': [50],     'Salary': [10000]})
pd.concat( [df1, df2], ignore_index=True) # VStack, df1 sorai után df2 sorai (NaN-ok)

df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [30, 40]})
df2 = pd.DataFrame({'Salary': [5000, 7000]})
pd.concat( [df1, df2], axis=1) # HStack, df1 oszlopai, utána df2 oszlopai, hstack, index szerint

df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [30, 40]})
df2 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Salary': [5000, 7000]})
pd.merge(df1, df2, on='Name') # inner join Col1 azonos értékei szerint, metszet

df1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [30, 40]})
df2 = pd.DataFrame({'Name': ['Alice', 'Cecil'], 'Salary': [5000, 10000]})
pd.merge(df1, df2, on='Name', how='outer')  # outer join, union

df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Cecil'], 'Position': ['boss', 'worker', 'boss'], 'Salary': [5000, 7000, 10000]})
df.groupby(['Position']).count() # azonos értékű Col sorok összevonása, többi megszámolása (nem null-ok)
df.groupby(['Position'])['Salary'].mean()
```
